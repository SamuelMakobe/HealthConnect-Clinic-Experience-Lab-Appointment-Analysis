## 3.5 Re-establish the Week 6 feature set

````python
# Loading the Week 6 processed dataset

import pandas as pd
import numpy as np

df = pd.read_csv("HealthConnect_Processed_Dataset.csv")

print("Dataset shape:", df.shape)
df.head()
````

````python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.columns.tolist()
````

````python
df["no_show"].value_counts()
````

````python
# Recreating the Week 6 engineered feature: previous no-show group

df["previous_no_show_group"] = np.where(df["previous_no_shows"] >= 3,"3+", df["previous_no_shows"].astype(str))

# Recreating the Week 6 interaction feature:
# booking lead-time group × previous no-show group

df["lead_no_show_interaction"] = (df["booking_lead_group"] + " | " + df["previous_no_show_group"]
)

# Checking the newly recreated features

print("previous_no_show_group:")
print(df["previous_no_show_group"].value_counts().sort_index())

print("\nlead_no_show_interaction:")
print(df["lead_no_show_interaction"].value_counts().sort_index())
````

````python
# Week 6 refined feature set

refined_features = ['gender','age','age_group','appointment_type','appointment_day',
                    'booking_lead_days','booking_lead_group','previous_appointments',
                    'previous_no_shows','previous_no_show_group','distance_to_clinic_km',
                    'distance_group','lead_no_show_interaction']

refined_features
````

````python
# Defining the numerical and categorical features

refined_numerical_features = ['age','booking_lead_days','previous_appointments',
                              'previous_no_shows','distance_to_clinic_km']

refined_categorical_features = ['gender','age_group','appointment_type','appointment_day',
                                'booking_lead_group','distance_group','previous_no_show_group',
                                'lead_no_show_interaction']

print("Numerical features:", refined_numerical_features)
print("Categorical features:", refined_categorical_features)
````

````python
# Verifying the refined feature lists

print("Total refined features:", len(refined_features))
print("Total numerical features:", len(refined_numerical_features))
print("Total categorical features:", len(refined_categorical_features))

print("\nAll features accounted for:",
      len(refined_numerical_features) + len(refined_categorical_features) == len(refined_features))
````

````python
# Creating the refined feature matrix and target

X_refined = df[refined_features].copy()
y_refined = df["no_show"].copy()

print("X_refined shape:", X_refined.shape)
print("y_refined shape:", y_refined.shape)
````

````python
# Checking the missing values in refined features

X_refined.isnull().sum()
````

````python
# Reviewing the refined categorical feature distributions

for column in refined_categorical_features:
    print(f"\n--- {column} ---")
    print(X_refined[column].value_counts())
````

## 3.6 Re-establish the Week 6 train-test split

````python
from sklearn.model_selection import train_test_split
X_train_refined, X_test_refined, y_train_refined, y_test_refined = train_test_split(X_refined,y_refined,
                                                                                    test_size=0.20,
                                                                                    random_state=42,
                                                                                    stratify=y_refined)

print("X_train_refined:", X_train_refined.shape)
print("X_test_refined:", X_test_refined.shape)
print("y_train_refined:", y_train_refined.shape)
print("y_test_refined:", y_test_refined.shape)
````

````python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
````

````python
# Creating the refined preprocessing pipeline

refined_preprocessor = ColumnTransformer(transformers=[("num", StandardScaler(), refined_numerical_features),
                                                       ("cat", OneHotEncoder(handle_unknown="ignore"), refined_categorical_features) ]
)

refined_preprocessor
````

## 3.7 Re-establish the Week 6 Logistic Regression Pipeline

````python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
````

````python
refined_logistic_model = Pipeline(steps=[("preprocessor", refined_preprocessor),
                                         ("classifier", LogisticRegression(max_iter=1000,random_state=42 ))])

refined_logistic_model
````

## 3.8 Train refined Logistic Regression model

````python
refined_logistic_model.fit(X_train_refined,y_train_refined)

print("Refined Logistic Regression model trained successfully.")
````

## 3.9 Generating predictions from the refined model

````python
y_pred_refined = refined_logistic_model.predict(X_test_refined)
y_prob_refined = refined_logistic_model.predict_proba(X_test_refined)[:, 1]

print("Predictions generated:", len(y_pred_refined))
print("Probability predictions generated:", len(y_prob_refined))
````

## 5.1 Error classification validation

The Week 6 candidate model produced false positives and false negatives that were examined during error analysis. In Week 7, these error categories are tested again using the reproduced test-set predictions.

The objective is to confirm that each test-set observation is consistently classified as a true negative, false positive, false negative, or true positive.

The error classification will then be used for the subsequent segment-level error testing.


````python
# Creating an error-analysis dataset from the Week 7 test-set predictions

error_analysis = X_test_refined.copy()

error_analysis["actual"] = y_test_refined.values
error_analysis["predicted"] = y_pred_refined

error_analysis["prediction_result"] = np.select(
    [
        (error_analysis["actual"] == 0) & (error_analysis["predicted"] == 0),
        (error_analysis["actual"] == 0) & (error_analysis["predicted"] == 1),
        (error_analysis["actual"] == 1) & (error_analysis["predicted"] == 0),
        (error_analysis["actual"] == 1) & (error_analysis["predicted"] == 1)
    ],
    [
        "True Negative",
        "False Positive",
        "False Negative",
        "True Positive"
    ],
    default="Unknown"
)

error_analysis["prediction_result"].value_counts()
````

### Result - Error classification

The Week 7 candidate model reproduced the Week 6 error classification exactly. The test set contained 247 true negatives, 216 false positives, 149 false negatives, and 336 true positives, matching the Week 6 results.

Test status: Pass

This confirms that the reproduced candidate model generates the same classification outcomes as the Week 6 model. The analysis can therefore proceed to testing whether the previously identified false-positive and false-negative patterns remain consistent across important data segments.


## 5.2 Booking lead-time error pattern validation

````python
# Testing whether the Week 6 error pattern by booking lead-time group is reproduced in the Week 7 candidate model

lead_error_test = pd.crosstab(error_analysis["booking_lead_group"],error_analysis["prediction_result"])

# Ensuring that all error categories are present and consistently ordered

lead_error_test = lead_error_test.reindex(columns=["True Negative","False Positive",
                                                   "False Negative","True Positive"
                                                  ],
                                          fill_value=0
)

# Calculating the total observations, total errors, and error rate

lead_error_test["Total"] = lead_error_test.sum(axis=1)

lead_error_test["Errors"] = (lead_error_test["False Positive"] + lead_error_test["False Negative"])

lead_error_test["Error Rate (%)"] = (lead_error_test["Errors"] / lead_error_test["Total"]* 100).round(2)

lead_error_test
````

### Result - Booking lead-time error pattern

The Week 7 testing reproduced the main booking lead-time error pattern identified during Week 6. The 15–30 day and 31–45 day groups continued to show the highest error rates at 47.37% and 46.99%, respectively. The 8–14 day group showed a moderate error rate of 36.70%, while the 0–7 day and 46–60 day groups had lower error rates of 24.37% and 29.22%.

Although the individual error counts differed slightly from the Week 6 results, the relative pattern across booking lead-time groups remained consistent. This supports the Week 6 finding that booking lead time is an important source of prediction error.

Test status: Pass

The result does not indicate that the candidate model is error-free. Instead, it validates that the previously identified booking lead-time weaknesses remain present and should continue to be considered during subsequent testing and refinement.

## 11.2 Create the additive risk score

````python
# Assigning ordinal scores to the existing booking lead-time groups

lead_score_map = {"0–7 days": 0,"8–14 days": 1,"15–30 days": 2,"31–45 days": 3,"46–60 days": 4}

df["booking_lead_score"] = df["booking_lead_group"].map(lead_score_map)


# Assigning ordinal scores to the existing previous no-show groups

previous_no_show_score_map = {"0": 0,"1": 1,"2": 2,"3+": 3}

df["previous_no_show_score"] = (df["previous_no_show_group"].map(previous_no_show_score_map))


# Creating the combined additive score

df["lead_no_show_score"] = (df["booking_lead_score"] + df["previous_no_show_score"])


# Checking the newly created variables

df[["booking_lead_group","booking_lead_score","previous_no_show_group",
    "previous_no_show_score","lead_no_show_score"]].head(10)
````

# 13. Limitations and risks

The Week 7 testing process identified several limitations and risks that should be considered before the HealthConnect model is used in a real operational environment.

## 13.1 Data limitations

The HealthConnect dataset is a fictional project dataset. Therefore, the observed relationships and model performance cannot be assumed to represent the behaviour of patients in a real healthcare environment.

The current evaluation is also based on the available dataset and its existing feature structure. Additional real-world variables may influence appointment attendance but are not represented in the current data.

## 13.2 Model performance limitations

The retained Logistic Regression model achieved moderate performance on the held-out test set:

Accuracy: 61.50%
Precision: 60.87%
Recall: 69.28%
F1-score: 64.80%
ROC-AUC: 66.47%

These results indicate that the model provides useful predictive information but does not classify appointments with sufficient accuracy to be treated as a definitive prediction system.

The model also produced 216 false positives and 149 false negatives. These errors could result in unnecessary attention being directed towards some patients while other patients at risk of missing appointments are not identified.

## 13.3 Segment performance risk

Model performance is not uniform across all appointment segments.

The most important variation identified during testing occurred across booking lead-time groups. Shorter booking lead-time groups showed relatively low recall, meaning that the model may fail to identify some no-show appointments within these groups.

This finding should therefore remain part of the testing and monitoring requirements for the next stage.

## 13.4 Validation Llimitations

The model demonstrated reasonable stability across the current training/test comparison and 5-fold cross-validation. However, these evaluations were performed using the same project dataset.

The model has not yet been independently validated using a separate external dataset or a temporally separated future dataset. Therefore, the current results do not establish how the model would perform on genuinely unseen future clinic data.

## 13.5 Probability and threshold limitations

The Week 7 testing did not fully evaluate probability calibration or alternative classification thresholds.

The current classification results therefore reflect the existing classification setup and should not be interpreted as demonstrating that the current threshold is optimal for HealthConnect's operational requirements.

Further testing could examine whether different thresholds provide a more appropriate balance between false positives and false negatives.

## 13.6 Operational risk

The model should not be used to make automatic decisions about patients or deny, cancel or restrict healthcare services.

If eventually incorporated into the HealthConnect workflow, its predictions should be treated as decision-support information that can help identify appointments for additional administrative attention.

Any operational implementation would require appropriate human oversight, validation and monitoring.

## 13.7 Interpretation risk

The relationships identified in the dataset are predictive associations and should not be interpreted as evidence that a particular factor causes a patient to miss an appointment.

For example, the relationship between booking lead time and no-show behaviour supports its use as a predictive feature, but does not by itself establish that changing booking lead time would cause attendance behaviour to change.

## 13.8 Cross-track integration risk

The Week 7 cross-track testing demonstrated that the Data Analytics recommendation could be translated into a modelling experiment. However, the tested additive lead_no_show_score did not improve the retained Logistic Regression model.

This demonstrates that an analytical finding can be valid without every proposed modelling representation improving predictive performance.

The result should therefore be treated as evidence for continued investigation rather than as evidence that the underlying Data Analytics findings are invalid.

## 13.9 Overall risk assessment

The main risks remaining at the end of Week 7 are:

1. Moderate overall predictive performance.
2. False-positive and false-negative classification errors.
3. Uneven performance across booking lead-time groups.
4. Lack of independent or temporal external validation.
5. Lack of complete probability calibration and threshold testing.
6. Limitations of the fictional project dataset.
7. The possibility that real-world variables are not represented in the current features.
8. The risk of treating predictive associations as causal relationships.
9. The need for human oversight if the model is eventually integrated into an operational workflow.

These limitations do not prevent the model from continuing to the next development stage, but they should be addressed through further testing and validation before any production deployment is considered.
