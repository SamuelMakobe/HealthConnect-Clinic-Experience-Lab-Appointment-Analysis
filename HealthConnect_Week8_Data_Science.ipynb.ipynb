# 3. Re-establishing the final candidate model

The Week 7 testing process confirmed that the Week 6 Refined Logistic Regression should remain the final candidate model.

For Week 8, the model will be reproduced from the established modelling workflow so that the final candidate can be formally documented, compared with the Week 5 baseline and prepared for final project integration.

The Week 8 notebook does not repeat the full Week 6 modelling process. Only the components necessary to reproduce and verify the final candidate model are re-established.

## 3.1 Load the processed dataset

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
# Load the processed HealthConnect dataset

df = pd.read_csv("HealthConnect_Processed_Dataset.csv")

print("Dataset shape:", df.shape)
df.head()
```

## 3.2 Define the final candidate features

```python
# Define the final candidate model features

refined_features = ['gender','age','age_group','appointment_type',
                    'appointment_day','booking_lead_days','booking_lead_group',
                    'previous_appointments','previous_no_shows','previous_no_show_group',
                    'distance_to_clinic_km','distance_group','lead_no_show_interaction'
                   ]

print("Number of refined features:", len(refined_features))
print("\nRefined features:")
for feature in refined_features:
    print("-", feature)
```

## 3.3 Checking that all features exist

```python
missing_features = [feature for feature in refined_features
                    if feature not in df.columns]

if len(missing_features) == 0:
    print("All 13 refined features are available.")
else:
    print("Missing features:")
    for feature in missing_features:
        print("-", feature)
```

## 3.4 Recreate the two engineered features

```python
df["previous_no_show_group"] = np.where(
    df["previous_no_shows"] >= 3,
    "3+",
    df["previous_no_shows"].astype(str)
)

df["previous_no_show_group"].value_counts().sort_index()
```

## 3.5 Recreate the interaction feature

```python
df["lead_no_show_interaction"] = (
    df["booking_lead_group"].astype(str)
    + " | "
    + df["previous_no_show_group"].astype(str)
)

print("Number of interaction groups:",
      df["lead_no_show_interaction"].nunique())

df["lead_no_show_interaction"].value_counts()
```

## 3.6 Rechecking the final 13 features

```python
missing_features = [feature for feature in refined_features
                    if feature not in df.columns
                   ]

if len(missing_features) == 0:
    print("All 13 refined features are available.")
else:
    print("Missing features:")
    for feature in missing_features:
        print("-", feature)
```

## 3.7  Defining X and y

```python
# Defining predictors and target

X = df[refined_features].copy()
y = df["no_show"].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())

print("\nTarget proportions:")
print(y.value_counts(normalize=True).round(4))
```

## 3.8 Creating the train/test split

```python
# Spliting the data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(X,y,
                                                    test_size=0.20,
                                                    random_state=42,
                                                    stratify=y)

print("Training set:", X_train.shape)
print("Test set:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTest target distribution:")
print(y_test.value_counts())
```

## 3.9 Identifying the numerical and categorical features

```python
numerical_features = ['age','booking_lead_days','previous_appointments',
                      'previous_no_shows','distance_to_clinic_km'
                     ]

categorical_features = ['gender','age_group','appointment_type',
                        'appointment_day','booking_lead_group',
                        'previous_no_show_group','distance_group',
                        'lead_no_show_interaction'
                       ]

print("Numerical features:", len(numerical_features))
print(numerical_features)

print("\nCategorical features:", len(categorical_features))
print(categorical_features)

print("\nTotal features:", len(numerical_features) + len(categorical_features))
```

## 3.10 Creating the preprocessing pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
```

```python
# Creating a preprocessing pipeline

preprocessor = ColumnTransformer(transformers=[("num", StandardScaler(), numerical_features),
                                               ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
                                              ])

print("Preprocessing pipeline created successfully.")
```

## 3.11 Building and training the final candidate model

```python
# Creating the final candidate Logistic Regression pipeline

final_model = Pipeline(steps=[("preprocessor", preprocessor),
                              ("classifier", LogisticRegression(max_iter=1000,
                                                                random_state=42))
                             ])

# Training the model
final_model.fit(X_train, y_train)

print("Final candidate model trained successfully.")
```

## 3.12 Generating the predictions

```python
y_pred_final = final_model.predict(X_test)
y_prob_final = final_model.predict_proba(X_test)[:, 1]

print("Predictions generated successfully.")
print("Number of predictions:", len(y_pred_final))
```

## 3.13 Evaluating the reproduced final candidate

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix)
```

```python
# Evaluating the reproduced final candidate model

final_accuracy = accuracy_score(y_test, y_pred_final)
final_precision = precision_score(y_test, y_pred_final)
final_recall = recall_score(y_test, y_pred_final)
final_f1 = f1_score(y_test, y_pred_final)
final_roc_auc = roc_auc_score(y_test, y_prob_final)

final_cm = confusion_matrix(y_test, y_pred_final)

print("Final candidate model performance")
print("---------------------------------")
print(f"Accuracy:  {final_accuracy:.4f}")
print(f"Precision: {final_precision:.4f}")
print(f"Recall:    {final_recall:.4f}")
print(f"F1 Score:  {final_f1:.4f}")
print(f"ROC-AUC:   {final_roc_auc:.4f}")

print("\nConfusion matrix:")
print(final_cm)
```

### Interpretation

The Week 6 Refined Logistic Regression was successfully reproduced in the Week 8 notebook. The reproduced model achieved an accuracy of 61.50%, precision of 60.87%, recall of 69.28%, F1 score of 64.80%, and ROC-AUC of 66.47%.

The confusion matrix produced 247 true negatives, 216 false positives, 149 false negatives, and 336 true positives. These results exactly match the final candidate performance confirmed during Week 7 testing.

This successful reproduction provides evidence that the final candidate model is reproducible and that the established modelling workflow has been carried forward consistently into Week 8. The model is therefore ready for final evaluation, comparison with the Week 5 baseline, business interpretation, and cross-track integration.

The final Refined Logistic Regression demonstrates moderate predictive performance and provides a useful proof-of-concept for identifying appointments with an elevated risk of no-show.

**Strengths**

The model achieved a recall of 69.28%, correctly identifying 336 of the 485 actual no-show appointments in the test set. This indicates that the model detects approximately seven out of every ten actual no-shows. Its ROC-AUC of 66.47% also indicates some ability to distinguish between appointments that result in attendance and those that result in a no-show.

**Weaknesses**

### 5.2 Integration of Data Analytics findings

The Data Analytics track provided validated findings that were used to support the interpretation and refinement of the Data Science model.

The strongest and most reliable analytical relationships were associated with booking lead time and previous no-show history.

#### Booking lead time

The Data Analytics findings showed a clear increase in no-show rate as booking lead time increased:

- 0–7 days: 27.81%
- 8–14 days: 33.55%
- 15–30 days: 43.21%
- 31–45 days: 53.82%
- 46–60 days: 67.69%

The relationship was statistically significant (χ² = 364.81, df = 4, p < 0.001), with a point-biserial correlation of 0.272.

These findings supported retaining both booking_lead_days and booking_lead_group in the final predictive model.

#### Previous no-Show history

The Data Analytics findings also showed increasing no-show rates as previous no-show history increased:

- 0 previous no-shows: 43.51%
- 1 previous no-show: 53.49%
- 2 previous no-shows: 59.36%
- 3 previous no-shows: 67.95%

The relationship was statistically significant (χ² = 80.71, df = 4, p < 0.001), with a point-biserial correlation of 0.126.

Very high previous no-show counts contained relatively few observations. This supported the grouped representation used by Data Science (0, 1, 2, 3+) while retaining the original previous_no_shows variable.

#### Cross-track interpretation

The Data Analytics results therefore independently supported the importance of booking lead time and previous no-show history within the Data Science workflow.

The analytical findings were used as evidence for feature interpretation and refinement rather than as evidence of causality. The observed relationships are correlational and should not be interpreted as demonstrating that longer booking lead times or previous no-shows directly cause future no-shows.

### 5.3 Final feature refinement decisions

The final feature decisions were based on evidence accumulated across the Week 6 refinement and Week 7 testing process rather than on adding features solely to increase model complexity.

#### Booking lead time × Previous no-show interaction

The final Week 6 candidate retained lead_no_show_interaction, which combines booking_lead_group and previous_no_show_group.

However, the validated Data Analytics findings indicated that booking lead time and previous no-show history showed stronger evidence as individual predictors, while their combined effect was better described as additive rather than strongly multiplicative.

The interaction feature is therefore retained as part of the already validated Week 6 candidate specification, but the project does not claim that the interaction itself is a dominant predictor or that it demonstrates a causal relationship.

#### Log-transformed booking lead time

During Week 7, an additional feature was tested:

booking_lead_log = log(1 + booking_lead_days)

The purpose was to determine whether a transformed representation of booking lead time could improve the model while preserving the strong relationship identified by Data Analytics.

The Week 7 model containing booking_lead_log achieved:

- Accuracy: 61.50%
- Precision: 60.91%
- Recall: 69.07%
- F1 Score: 64.73%
- ROC-AUC: 66.52%

Compared with the retained Week 6 candidate, the transformation produced only marginal changes. Recall decreased from 69.28% to 69.07% and F1 score decreased from 64.80% to 64.73%, while ROC-AUC increased only slightly from 66.47% to 66.52%.

Segment-level testing also showed no consistent improvement across the booking lead-time groups.

For this reason, booking_lead_log was rejected and the original Week 6 feature specification was retained.

#### Final feature decision

No additional feature was added to the final candidate during Week 8. The established 13-feature specification was retained because it had already progressed through feature refinement, model comparison, cross-validation, error analysis, segment testing and Week 7 re-testing.

This decision prioritises evidence, reproducibility and continuity over unnecessary model complexity.

Two main predictive models were evaluated during the model-development process: the Refined Logistic Regression and Random Forest.

The Refined Logistic Regression produced higher performance across all five reported test-set metrics. The largest practical difference was observed in recall, where Logistic Regression identified 69.28% of actual no-shows compared with 62.68% for Random Forest.

# 7. Final model limitations, risks and appropriate use

The final Refined Logistic Regression provides a tested predictive prototype for identifying appointments with an elevated likelihood of no-show. However, its outputs must be interpreted within the limitations of the HealthConnect project and the observed model performance.

## 7.1 Model limitations

**Moderate predictive performance**

The final model achieved an accuracy of 61.50%, recall of 69.28%, F1 score of 64.80% and ROC-AUC of 66.47%. These results indicate useful predictive signal, but they are not sufficiently strong to treat individual predictions as definitive outcomes.

**False-positive predictions**

The model incorrectly classified 216 patients who actually attended as no-shows, corresponding to a false-positive rate of 46.65%.

If model predictions were used directly for operational decisions, this could result in unnecessary interventions being directed toward patients who would have attended their appointments.

**False-negative predictions**

The model failed to identify 149 actual no-shows, corresponding to a false-negative rate of 30.72%.

Therefore, the model cannot identify every appointment that will result in a no-show.

**Dataset limitations**

HealthConnect is a fictional educational dataset used for the AnalystLab Africa Experience Lab. The model has therefore not been validated using an independent real-world healthcare population.

Performance observed in this project should not be assumed to generalise to real clinics, different patient populations or future appointment periods.

**Limited external and temporal validation**

The model was evaluated using a train/test split and stratified cross-validation within the available dataset. Independent external validation and prospective temporal validation have not been performed.

**Correlation rather than causation**

Relationships identified in the Data Analytics and Data Science tracks represent statistical associations within the available data. They do not demonstrate that variables such as booking lead time or previous no-show history directly cause future no-shows.

**Classification threshold**

The reported classification results use the established model decision threshold. Alternative thresholds may change the balance between false positives and false negatives. A formal threshold-optimisation analysis has not been completed in the current project.

**Probability calibration**

The model produces predicted no-show probabilities, but a dedicated probability-calibration assessment has not been performed. These probabilities should therefore not automatically be interpreted as precisely calibrated real-world risk estimates.

## 7.2 Key risks

The primary risk is over-reliance on the model's predictions. A patient classified as a likely no-show may still attend, while a patient predicted to attend may still miss the appointment.

Using predictions as the sole basis for decisions could therefore lead to inappropriate or inefficient interventions.

There is also a generalisation risk because the model has not been externally validated on operational healthcare data. Performance could change when applied to a different population, clinic or time period.

## 7.3 What the model can and cannot do

The model can:

- identify statistical patterns associated with appointment no-shows within the HealthConnect dataset;
- estimate a relative no-show probability using the established predictor set;
- classify appointments as predicted Attended or predicted No-Show using the established decision threshold;
- support analysis of appointments that may warrant additional attention;
- demonstrate an end-to-end predictive modelling workflow for the HealthConnect project.

The model cannot:

- guarantee whether an individual patient will attend or miss an appointment;
- establish causal relationships between predictor variables and no-show behaviour;
- replace human or operational decision-making;
- be assumed to perform equally well in real healthcare environments;
- be considered production-ready without further validation and implementation testing.

## 7.4 Appropriate use

Within the HealthConnect project, the model should be presented as a decision-support prototype rather than an automated decision-making system.

Its value lies in demonstrating how historical appointment information can contribute to identifying elevated no-show risk while also showing the importance of model validation, error analysis and responsible interpretation.

Before any real-world implementation, further work would be required, including external and temporal validation, threshold optimisation, probability calibration, operational testing and assessment of the consequences of model errors.
