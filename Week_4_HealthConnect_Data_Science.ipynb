## 3. Proposed target variable

### 3.1 Target variable

The proposed target variable is appointment_outcome. The variable records the final status of each appointment and contains three categories: Attended, No-Show, and Cancelled.

The primary outcome of interest is No-Show, because the objective is to investigate whether an appointment can be identified as being at risk of being missed.

### 3.2 Handling appointment outcomes

For the initial no-show prediction problem, the target will be treated as a binary classification outcome. Appointments recorded as No-Show will represent the positive class, while appointments recorded as Attended will represent the negative class.

Appointments recorded as Cancelled will be excluded from the initial prediction target because a cancellation represents a different appointment outcome from a patient failing to attend. The treatment of cancelled appointments can be revisited in a later stage if the project requires a separate cancellation prediction or a broader non-attendance analysis.

No records will be removed or target values transformed during Week 4 and this represents the proposed approach for subsequent modelling stages.

## 4. Potential features

### 4.1 Potential input features

Potential input features were identified based on their relevance to patient characteristics, appointment details, booking behaviour, previous appointment history, reminder information, and accessibility or operational factors. The initial candidate features are gender, age, age_group, appointment_type, appointment_day, appointment_time, booking_lead_days, previous_appointments, previous_no_shows, reminder_sent, reminder_channel, distance_to_clinic_km, and waiting_time_minutes.

Identifier variables such as appointment_id and patient_id are not proposed as predictive features because they primarily identify appointments or patients. The appointment_outcome variable is reserved as the target. The booking and appointment dates may be considered later for deriving useful time-related features, but no feature engineering is proposed at the Week 4 stage.

### 4.2 Feature considerations

Some potential features may require further assessment before they are used for modelling. In particular, age_group is derived from age, while booking_date and appointment_date may be used to derive additional time-related information. Variables such as waiting_time_minutes should also be assessed for potential information leakage because their availability at the time of prediction must be considered.

These considerations will be investigated during the subsequent modelling stage rather than being resolved during Week 4.

## 5. Initial modelling approach

### 5.1 Proposed modelling strategy

The proposed machine-learning approach is a supervised binary classification problem. The model will use historical appointment information to predict whether a scheduled appointment is likely to result in a No-Show or an Attended outcome.

The initial modelling stage will involve preparing the selected features, separating the data into training and testing sets, and comparing suitable classification algorithms. The final model selection will be based on appropriate evaluation metrics rather than accuracy alone.

### 5.2 Proposed data preparation

The selected features will require appropriate preprocessing before they can be used for machine-learning. This is expected to include handling missing values, converting categorical variables into a suitable numerical representation, and ensuring that date-related information is represented appropriately where required.

The proposed binary target will also need to be encoded into a numerical form during the modelling stage. Any preprocessing decisions will be applied using only information available from the appropriate training data to reduce the risk of data leakage.

### 5.3 Model evaluation

The model will be evaluated using classification metrics that provide insight into its ability to identify no-show appointments. These may include precision, recall, F1-score, and a confusion matrix, with particular attention given to recall for the No-Show class. Accuracy may also be reported but will not be used as the sole measure of model performance.

The final evaluation approach will be confirmed during the modelling stage after the class distribution and model performance have been assessed.

## 6. Key modelling considerations

### 6.1 Missing data

Missing values will need to be addressed before model development. The treatment should consider the meaning of the missing values rather than applying the same method to all variables. In particular, the missing values in reminder_channel require careful consideration because the Data Dictionary indicates that None may occur when no reminder was sent. The appropriate treatment of missing values will therefore be determined during the data preparation stage.

### 6.2 Class distribution

The appointment outcomes are distributed across No-Show (48.46%), Attended (46.28%), and Cancelled (5.26%). Since the proposed prediction problem focuses on No-Show versus Attended outcomes, cancelled appointments will be treated separately as described in Section 3.2. The resulting class distribution should be assessed again during model development to determine whether any class imbalance requires consideration.

### 6.3 Data leakage

Data leakage must be considered when selecting and preparing features. Variables should only be used if the information would be available at the time a no-show prediction is made. In particular, waiting_time_minutes requires further assessment because information about actual or estimated waiting time may not be available at the prediction point. The timing and availability of other variables will also be reviewed during model development to reduce the risk of leakage.

### 6.4 Limitations and risks

The dataset is synthetic and anonymised, so the findings may not fully represent real world patient behaviour or clinic operations. The dataset also contains missing values in selected variables, which will require appropriate treatment during model development. In addition, the treatment of cancelled appointments and the availability of certain variables at the prediction point may affect the final modelling approach. These limitations will be considered when interpreting model results and assessing the potential usefulness of the proposed solution.
